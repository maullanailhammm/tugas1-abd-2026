# Milestone 1: Data Profiling

Profil awal dataset besar Indonesia menggunakan Polars LazyFrame dan DuckDB. Isi sumber serta deskripsi dataset pada `data/README.md` sebelum menjalankan notebook ini.

```python
from pathlib import Path

import duckdb
import polars as pl

PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "data.csv"
if not DATA_PATH.exists():
    raise FileNotFoundError(f"Dataset tidak ditemukan: {DATA_PATH}")
print(f"Dataset: {DATA_PATH.name}")
print(f"Lokasi: {DATA_PATH}")
print(f"Ukuran file: {DATA_PATH.stat().st_size / 1_000_000:.1f} MB")
```

## Profil Polars

`scan_csv` hanya membangun rencana query; file baru dibaca ketika `collect()` dipanggil.

```python
lazy_data = pl.scan_csv(DATA_PATH, infer_schema_length=10000)
print(lazy_data.collect_schema())
ringkasan = lazy_data.select([
    pl.len().alias("jumlah_baris"),
    pl.all().null_count().name.suffix("_null"),
]).collect()
ringkasan
```

```python
contoh_data = lazy_data.head(10).collect()
contoh_data
```

## Rentang waktu dan sumber berita

Kolom `date` disimpan sebagai teks dengan zona waktu `+00`. Bagian waktu hingga detik diparse sebagai waktu UTC untuk menghitung rentang data.

```python
profil_temporal = lazy_data.select(
    pl.col("date")
    .str.slice(0, 19)
    .str.strptime(pl.Datetime, format="%Y-%m-%d %H:%M:%S", strict=True)
    .alias("date_parsed")
).select([
    pl.col("date_parsed").min().alias("tanggal_awal"),
    pl.col("date_parsed").max().alias("tanggal_akhir"),
]).collect()

profil_sumber = lazy_data.select(
    pl.col("source").drop_nulls().unique().sort().alias("sumber_berita")
).collect()

print("Rentang waktu data:")
print(profil_temporal)
print(f"Jumlah sumber berita: {profil_sumber.height}")
print(f"Daftar sumber berita: {profil_sumber['sumber_berita'].to_list()}")
```

## Profil DuckDB

DuckDB membaca file CSV langsung dari disk dan menghasilkan ringkasan statistik setiap kolom.

```python
connection = duckdb.connect()
profil_duckdb = connection.execute(
    "SUMMARIZE SELECT * FROM read_csv_auto(?)", [str(DATA_PATH)]
).pl()
profil_duckdb
```
